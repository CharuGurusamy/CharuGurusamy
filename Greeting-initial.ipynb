```python
import os
import google.generativeai as genai
from PIL import Image

def run_gemini_agent(image_path):
  """Agent that transcribes text and analyses handwriting using gemini"""
  if not os.path.exists(image_path):
    raise FileNotFoundError(f"Image doesn't exist{image_path}")

  # Configure the API key explicitly
  genai.configure(api_key=os.environ.get('GEMINI_API_KEY'))

  print("🔄 Opening and preparing image...")
  img = Image.open(image_path)

  system_instruction_text = (
      "1. You are an expert in analysing the image and converting it into text\\n"
      "2. Accurately transcribe every word from the handwriting in the image \\n"
      "3. Analyse the handwriting of the user e.g., slant, spacing, pressure, legibility) and provide text for the same"
  )

  print("AI is thinking")

  try:
    # Instantiate the GenerativeModel with the model name and system instruction
    model = genai.GenerativeModel(
        model_name='gemini-3.5-flash', # Changed model name
        system_instruction=system_instruction_text
    )

    # Call generate_content on the model instance
    response = model.generate_content(
        contents=[img, 'Please transcribe the image to text, make no mistakes'],
        generation_config=genai.GenerationConfig(temperature=0.2)
    )
    print(response.text)

  except Exception as e :
    print(f"error occured,{e}")

image_file_path = "/Greetings/dayspringpen-teacher-handwriting.webp"
run_gemini_agent(image_file_path)
```
